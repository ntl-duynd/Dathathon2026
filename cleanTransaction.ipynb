```python
# Trong số các khách hàng có nhiều hơn một đơn hàng, trung vị số ngày giữa hai lần
# mua liên tiếp (inter-order gap) xấp xỉ là bao nhiêu? (Tính từ orders.csv)
dfCustomermore2Order = dfOrder.groupby("customer_id")["order_date"].value_counts().reset_index().sort_values(by = ["customer_id", "order_date"])
```

```python
listDay = []
for i in dfCustomermore2Order2.index:
    #     # print(dfCustomermore2Order2[i])
    #     for j in range (1, len(dfCustomermore2Order2[i]), 1):
    #         res = (dfCustomermore2Order2[i][j] - dfCustomermore2Order2[i][j-1]).days
    #         print(i, res)
    if len(dfCustomermore2Order2[i]) >= 2:
        for j in range (1, len(dfCustomermore2Order2[i]), 1):
            # print(i ,dfCustomermore2Order2[i][j])
            res = (dfCustomermore2Order2[i][j] - dfCustomermore2Order2[i][j-1]).days
            if res not in listDay:
                listDay.append(res)
```

```python
import pandas as pd

dfOrder['gap'] = dfOrder.groupby('customer_id')['order_date'].diff().dt.days

# lọc khách có >1 đơn (gap != null)
gaps = dfOrder['gap'].dropna()

# median
median_gap = gaps.median()

print(median_gap)
```
